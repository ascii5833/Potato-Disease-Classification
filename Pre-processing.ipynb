```python
import pandas as pd 
import numpy as np
```

```python
sheet=pd.ExcelFile("Train_data.xlsx")
sheet.sheet_names
```

```python
df1=sheet.parse('Sheet1')
df1.head()
```

```python
df1.columns
```

```python
df2=sheet.parse('Sheet2')
df2.columns
```

```python
df3=sheet.parse('Sheet3')
df3.columns
```

```python
print(len(df1.columns),len(df2.columns),len(df3.columns))
```

```python
df1=df1.loc[:,~df1.columns.str.contains('^Unnamed')]
df1.head()
```

<h2>Adding prediction vector</h2>

```python
df1['class']=[0 for ele in df1['Autocorrelation']]
df1.head()
```

```python
df2['class']=[1 for ele in df2['Autocorrelation']]
df2.head()
```

```python
df3['class']=[2 for ele in df3['Autocorrelation']]
df3.head()
```

<h2> Removing unwanted columns</h2>

```python
df1=df1.drop(labels='Filename',axis=1)
df2=df2.drop(labels='Filename',axis=1)
df3=df3.drop(labels='Filename',axis=1)
df1.head()
```

<h2> Concatenate Dataframes</h2>

```python
df_main=pd.concat([df1,df2,df3],axis=0)
df_main.head()
```

```python
df_main.describe()
```

```python
df_main.isna().sum()
```

```python
#no null values creating a csv file from this
df_main.to_csv('./Data/train_data.csv',index=False)
```
