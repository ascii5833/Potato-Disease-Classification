```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Data/test_data.csv')
df.head()
```

```python
def transform(x):
    if x==0 or x==1:
        return 1
    return 0
df['class']=np.array(list(map(lambda x:transform(x),df['class'])))
```

```python
df.head()
```

```python
df['class'].value_counts()
```

```python
df.to_csv('./Data/test_detection.csv',index=False)
```
