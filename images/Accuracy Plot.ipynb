```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Accuracies Potato Disease Classification - Sheet1.csv")
df.head()
```

<h2>Plotting graph</h2>

```python
g=sns.catplot(data=df,x="Method",y="Accuracy",kind="bar")
g.set(xlabel='Classification Algorithm', ylabel='Accuracy')
g.fig.set_size_inches(10,7)
plt.xlabel("Classification Algorithm",size=16)
plt.ylabel("Accuracy",size=16)
plt.title("Accuracy for various classification algorithms using all features",size=16)
plt.savefig("./histoAccuracies.png",dpi=1200,bbox_inches="tight")
```

```python
g=sns.catplot(data=df,x="Method",y="Special Accuracy",kind="bar")
g.set(xlabel='Classification Algorithm', ylabel='Accuracy')
g.fig.set_size_inches(10,7)
plt.xlabel("Classification Algorithm",size=16)
plt.ylabel("Accuracy",size=16)
plt.title("Accuracy for various classification algorithms using researched features",size=16)
plt.savefig("./histoResearchFeaturesAccuracies.png",dpi=1200,bbox_inches="tight")
```
